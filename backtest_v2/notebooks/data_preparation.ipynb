```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read CSV with semicolon separator and comma decimal
df_totret = pd.read_csv("../data/TOTRET_DAILY.csv", sep=';', decimal=',', parse_dates=['Date'], dayfirst=True, na_values=['#N/A'])

df_totret.set_index('Date', inplace=True)
```

```python
# Read market cap and compute per-date portfolio weights
df_mcap = pd.read_csv("../data/MCAP_DAILY.csv", sep=';', decimal=',', parse_dates=['Date'], dayfirst=True, na_values=['#N/A'])
# Use Date as index
df_mcap.set_index('Date', inplace=True)
# Ensure numeric values (coerce malformed entries to NaN)
df_mcap = df_mcap.apply(pd.to_numeric, errors='coerce')
# Total market cap per date
total_mcap = df_mcap.sum(axis=1)
# Weights: each asset mcap divided by total mcap on the same date
weights = df_mcap.div(total_mcap, axis=0)
# Inspect the first rows of weights
weights.head()
```

```python
stock_ret = df_totret.pct_change()
index_ret = (weights.shift(1) * stock_ret).sum(axis=1)
index_tr = (1 + index_ret).cumprod() * 100
index_tr.plot()
```

```python
mva50  = df_totret.rolling(50,  min_periods=1).mean()
mva100 = df_totret.rolling(100, min_periods=1).mean()
mva200 = df_totret.rolling(200, min_periods=1).mean()
```

```python
features = pd.concat(
    {"tr": df_totret, "mva50": mva50, "mva100": mva100, "mva200": mva200},
    axis=1
)
# access like:
# features["mva200"]["EQNR.OL"]
```

```python


# 1) Read semicolon-separated file
mem_raw = pd.read_csv("../data/ON_INDEX.csv", sep=";", engine="python")

# 2) Rename first column to 'date' (it should be DATES)
mem_raw = mem_raw.rename(columns={mem_raw.columns[0]: "date"})

# 3) Parse dates (dayfirst because you have 04.03.2006)
mem_raw["date"] = pd.to_datetime(mem_raw["date"], dayfirst=True, errors="coerce")
mem_raw = mem_raw.dropna(subset=["date"]).set_index("date").sort_index()

# 4) Drop the 'id()' row if present (often first row after header)
#    Your screenshot shows a row containing 'id()' values.
mask_id_row = mem_raw.apply(lambda row: row.astype(str).str.contains(r"id\(\)", case=False, na=False)).any(axis=1)
mem_raw = mem_raw.loc[~mask_id_row]

# 5) Convert to 0/1 membership: cell non-empty => 1 else 0
mem_str = mem_raw.astype("string")
membership_m = (mem_str.notna() & (mem_str != "")).astype(int)

# 6) Expand monthly to daily and align to your features index
membership_d = membership_m.reindex(features.index, method="ffill").fillna(0).astype(int)

# Align columns to features tickers (important)
tickers = features["tr"].columns
membership_d = membership_d.reindex(columns=tickers).fillna(0).astype(int)

# 7) Filter features: keep values only when member==1
filtered_fields = {}
for field in features.columns.get_level_values(0).unique():
    filtered_fields[field] = features[field].where(membership_d.eq(1))

features_osebx = pd.concat(filtered_fields, axis=1).sort_index(axis=1)

print(membership_m.shape, membership_d.shape, features_osebx.shape)
```

# MVA Order Scores on OSEBX

This notebook restricts the moving-average order-score study to the daily `ON_INDEX` OSEBX universe. The 50/100/200-day moving averages now require full windows and are computed from in-index history only. It searches all six order-regime subsets on the actual portfolio objects that are later reported, selects one best method for equal weighting and one for cap weighting, and then reports monthly CAPM alpha significance against the OSEBX cap-weight proxy using a 3% annual risk-free rate. Because the subsets are selected in sample, the reported alpha p-values are exploratory rather than confirmatory.


```python
import importlib
import sys
from pathlib import Path
sys.path.insert(0, str((Path.cwd() / '../src').resolve()))

import matplotlib.pyplot as plt
import pandas as pd

import mva_order_utils
importlib.reload(mva_order_utils)

RF_MONTHLY = mva_order_utils.RF_MONTHLY
compute_order_score = mva_order_utils.compute_order_score
load_mva_inputs = mva_order_utils.load_mva_inputs
summarize_best_order_methods = mva_order_utils.summarize_best_order_methods

pd.set_option("display.float_format", lambda x: f"{x:,.6f}")
pd.set_option("display.max_colwidth", None)
```

```python
inputs = load_mva_inputs()
features = inputs["features"]
weights_wide = inputs["weights_wide"]
on_index_m = inputs["on_index_m"]
market_monthly_ret = inputs["market_monthly_ret"]

print("TR panel:", inputs["tr_daily"].shape, inputs["tr_daily"].index.min().date(), "to", inputs["tr_daily"].index.max().date())
print("ON_INDEX monthly:", on_index_m.shape, on_index_m.index.min().date(), "to", on_index_m.index.max().date())
print("Average OSEBX members:", round(on_index_m.sum(axis=1).mean(), 1))
print("Annual risk-free rate:", "3.00%")
print("Monthly risk-free rate:", f"{RF_MONTHLY:.6%}")
```

```python
stock = "EQNR NO Equity"
fig, ax = plt.subplots(figsize=(11, 5))
features["tr"][stock].dropna().plot(ax=ax, lw=1.3, label="TR")
features["mva50"][stock].dropna().plot(ax=ax, lw=1.0, label="MVA50")
features["mva100"][stock].dropna().plot(ax=ax, lw=1.0, label="MVA100")
features["mva200"][stock].dropna().plot(ax=ax, lw=1.0, label="MVA200")
ax.set_title(f"{stock} (values only when in OSEBX)")
ax.set_ylabel("Level")
ax.legend()
ax.grid(True, alpha=0.25)
plt.show()
```

```python
order_score = compute_order_score(features)
display(order_score.dropna(how="all").tail())
display(order_score.stack().rename("order_score").value_counts().sort_index().to_frame("n_obs"))
```

```python
results = summarize_best_order_methods(
    features=features,
    weights_wide=weights_wide,
    on_index_m=on_index_m,
    market_monthly_ret=market_monthly_ret,
    cap=1.0,
    min_months=60,
    rf_monthly=RF_MONTHLY,
    selection_metric="Sharpe",
)

print("Best order subset for equal weighting:", results["best_equal_subset"])
print("Best order subset for cap weighting:", results["best_cap_subset"])
```

```python
search = results["search_results"].copy()
display(search.loc[search["weighting"].eq("equal")].sort_values("Sharpe", ascending=False).head(15))
display(search.loc[search["weighting"].eq("cap")].sort_values("Sharpe", ascending=False).head(15))
```

```python
print("Best-method summary with monthly CAPM alpha tests")
print("Subset selection is in sample and based on portfolio Sharpe, so the alpha p-values below are exploratory.")
display(results["summary"])
```

```python
equal_strategy = results["equal_strategy"]
cap_strategy = results["cap_strategy"]
ret = mva_order_utils.compute_clean_returns(features, cap=1.0)
equal_market_monthly_ret = ((1.0 + mva_order_utils.index_returns_equal(ret)).resample("ME").prod() - 1.0).rename("equal_market")
cap_market_equity = (1.0 + market_monthly_ret).cumprod()

fig, ax = plt.subplots(figsize=(11, 5))
eq_market = (1.0 + equal_market_monthly_ret.reindex(equal_strategy["monthly_equity"].index)).cumprod()
eq_market = eq_market / eq_market.iloc[0]
equal_strategy["monthly_equity"].plot(ax=ax, lw=1.8, label=f"Equal best {results['best_equal_subset']}")
eq_market.plot(ax=ax, lw=1.3, color="black", linestyle="--", label="OSEBX equal-weight index")
ax.set_title("Best Equal-Weight Order Strategy vs OSEBX Equal-Weight Index")
ax.set_ylabel("Equity, start = 1")
ax.grid(True, alpha=0.25)
ax.legend()
plt.show()

fig, ax = plt.subplots(figsize=(11, 5))
cap_market = cap_market_equity.reindex(cap_strategy["monthly_equity"].index)
cap_market = cap_market / cap_market.iloc[0]
cap_strategy["monthly_equity"].plot(ax=ax, lw=1.8, color="tab:blue", label=f"Cap best {results['best_cap_subset']}")
cap_market.plot(ax=ax, lw=1.4, color="black", linestyle="--", label="OSEBX cap-weight index")
ax.set_title("Best Cap-Weight Order Strategy vs OSEBX Cap-Weighted Index")
ax.set_ylabel("Equity, start = 1")
ax.grid(True, alpha=0.25)
ax.legend()
plt.show()
```

```python
alpha_table = pd.DataFrame(
    {
        "equal_weight_best": results["equal_capm"],
        "cap_weight_best": results["cap_capm"],
    }
)
display(alpha_table)
```

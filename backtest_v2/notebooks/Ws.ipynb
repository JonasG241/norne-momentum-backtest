# Winners-Only OSEBX Momentum Strategy

This notebook implements a long-only winners strategy using the same no-lookahead timing and OSEBX membership constraints as `WsLs_new.ipynb`, but without shorting losers. Returns are therefore interpretable as returns on an initial long-only investment.

Methodology:

- Rank only OSEBX members at each formation month.
- Use momentum signal data only through `t - skip`.
- Select the top `q` winners.
- Check that selected names are still OSEBX members on the next trading day.
- With close-only data, trade at that next trading day's close and start earning returns from the following close-to-close return.
- Hold each cohort as buy-and-hold for `K` months, allowing weights to drift.
- Test CAPM alpha from monthly returns: `R_winners,t - rf_t = alpha + beta * (R_m,t - rf_t) + eps_t`.

The risk-free rate is set to 3% per year and converted to the compounded monthly equivalent for CAPM tests.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math

pd.set_option("display.float_format", lambda x: f"{x:,.6f}")
```

```python
tr_daily = pd.read_csv(
    "../data/TOTRET_DAILY.csv",
    sep=";",
    decimal=",",
    parse_dates=["Date"],
    dayfirst=True,
    na_values=["#N/A"],
)
tr_daily = tr_daily.set_index("Date").sort_index()
tr_daily = tr_daily.loc[:, tr_daily.notna().any(axis=0)]

membership_m = pd.read_csv("../data/membership_m.csv", parse_dates=["date"])
membership_m = membership_m.set_index("date").sort_index()
membership_m = membership_m.apply(pd.to_numeric, errors="coerce").fillna(0).astype(int)
membership_m = membership_m.reindex(columns=tr_daily.columns).fillna(0).astype(int)

weights_wide = pd.read_csv("../data/weights_wide.csv", index_col=0, parse_dates=True)
weights_wide = weights_wide.reindex(columns=tr_daily.columns).fillna(0.0)

def load_on_index_membership(path: str, columns: pd.Index) -> pd.DataFrame:
    on_index_raw = pd.read_csv(path, sep=";", engine="python")
    on_index_raw = on_index_raw.rename(columns={on_index_raw.columns[0]: "date"})
    on_index_raw = on_index_raw.loc[~on_index_raw["date"].astype(str).str.fullmatch("DATES", case=False, na=False)]
    on_index_raw["date"] = pd.to_datetime(on_index_raw["date"], format="%d.%m.%Y", dayfirst=True, errors="coerce")
    on_index_raw = on_index_raw.dropna(subset=["date"]).set_index("date").sort_index()
    mask_id_row = on_index_raw.apply(lambda row: row.astype(str).str.contains(r"id\(\)", case=False, na=False)).any(axis=1)
    on_index_raw = on_index_raw.loc[~mask_id_row]
    on_index_str = on_index_raw.astype("string")
    on_index_m = (on_index_str.notna() & (on_index_str != "")).astype(int)
    return on_index_m.reindex(columns=columns).fillna(0).astype(int)

on_index_m = load_on_index_membership("../data/ON_INDEX.csv", tr_daily.columns)

print("TR daily:", tr_daily.shape, tr_daily.index.min().date(), "to", tr_daily.index.max().date())
print("Membership monthly:", membership_m.shape, membership_m.index.min().date(), "to", membership_m.index.max().date())
print("Weights:", weights_wide.shape)
```

```python
def expand_membership_to_daily(membership: pd.DataFrame, daily_index: pd.DatetimeIndex) -> pd.DataFrame:
    return membership.sort_index().reindex(daily_index, method="ffill").fillna(0).astype(int)


def month_end_last_trading_dates(daily_index: pd.DatetimeIndex) -> pd.Series:
    s = pd.Series(daily_index, index=daily_index)
    return s.resample("ME").max().dropna()


def rank_top_tail(signal_xs: pd.DataFrame, q: float) -> pd.DataFrame:
    winners = pd.DataFrame(False, index=signal_xs.index, columns=signal_xs.columns)
    for date, row in signal_xs.iterrows():
        valid = row.dropna()
        n = len(valid)
        if n == 0:
            continue
        n_tail = max(1, int(np.floor(q * n)))
        ordered = valid.sort_values(kind="mergesort")
        winners.loc[date, ordered.index[-n_tail:]] = True
    return winners


def normalize_long_only(weights: pd.DataFrame) -> pd.DataFrame:
    denom = weights.sum(axis=1).replace(0, np.nan)
    return weights.div(denom, axis=0).fillna(0.0)


def max_drawdown(equity: pd.Series) -> float:
    return float((equity / equity.cummax() - 1).min())


def summarize_returns(r: pd.Series, periods_per_year: int = 252) -> pd.Series:
    r = pd.Series(r).dropna()
    equity = (1 + r).cumprod()
    years = len(r) / periods_per_year
    vol = r.std() * np.sqrt(periods_per_year)
    sharpe = (r.mean() / r.std()) * np.sqrt(periods_per_year) if r.std() > 0 else np.nan
    cagr = equity.iloc[-1] ** (1 / years) - 1 if years > 0 and len(equity) else np.nan
    return pd.Series({
        "CAGR": cagr,
        "Vol": vol,
        "Sharpe": sharpe,
        "MaxDD": max_drawdown(equity) if len(equity) else np.nan,
        "Periods": len(r),
        "FinalEquity": equity.iloc[-1] if len(equity) else np.nan,
    })
```

```python
def winners_only_backtest(
    tr_daily: pd.DataFrame,
    membership_m: pd.DataFrame,
    J: int = 12,
    K: int = 12,
    q: float = 0.10,
    skip: int = 1,
    min_names: int = 10,
    require_member_entire_formation: bool = False,
    fill_missing_prices_while_held: bool = True,
) -> dict:
    """Long-only overlapping buy-and-hold winner cohorts."""
    tr_daily = tr_daily.sort_index()
    membership_m = membership_m.sort_index().reindex(columns=tr_daily.columns).fillna(0).astype(int)
    membership_d = expand_membership_to_daily(membership_m, tr_daily.index)

    tr_for_returns = tr_daily.ffill() if fill_missing_prices_while_held else tr_daily
    ret_d_raw = tr_for_returns.pct_change(fill_method=None)
    ret_d = ret_d_raw.fillna(0.0) if fill_missing_prices_while_held else ret_d_raw.copy()

    tr_m = tr_daily.resample("ME").last()
    member_m = membership_d.resample("ME").last().reindex(tr_m.index).fillna(0).astype(bool)

    signal = tr_m.shift(skip).pct_change(periods=J, fill_method=None)
    eligible = member_m & signal.notna()

    if require_member_entire_formation:
        formation_member = (
            member_m.astype(int)
            .shift(skip)
            .rolling(J, min_periods=J)
            .min()
            .eq(1)
        )
        eligible &= formation_member

    eligible = eligible.where(eligible.sum(axis=1).ge(min_names), False)
    signal_xs = signal.where(eligible)
    winners = rank_top_tail(signal_xs, q=q)

    formed_weights = normalize_long_only(winners.astype(float))

    last_trade_by_month = month_end_last_trading_dates(tr_daily.index)
    trading_days = pd.Series(tr_daily.index, index=tr_daily.index)
    entry_member = pd.DataFrame(False, index=formed_weights.index, columns=formed_weights.columns)
    entry_price_ok = pd.DataFrame(False, index=formed_weights.index, columns=formed_weights.columns)
    entry_possible = pd.Series(False, index=formed_weights.index)
    entry_dates = pd.Series(pd.NaT, index=formed_weights.index, dtype="datetime64[ns]")

    for month_end in formed_weights.index:
        if month_end not in last_trade_by_month.index:
            continue
        formation_trade_date = last_trade_by_month.loc[month_end]
        next_days = trading_days.loc[trading_days.index > formation_trade_date]
        if next_days.empty:
            continue
        entry_date = next_days.iloc[0]
        entry_possible.loc[month_end] = True
        entry_dates.loc[month_end] = entry_date
        entry_member.loc[month_end] = membership_d.loc[entry_date].astype(bool)
        entry_price_ok.loc[month_end] = tr_daily.loc[entry_date].notna() & ret_d.loc[entry_date].notna()

    entry_ok = entry_member & entry_price_ok
    opened_weights = normalize_long_only(formed_weights.where(entry_ok, 0.0))

    dates = tr_daily.index
    columns = tr_daily.columns
    date_pos = pd.Series(np.arange(len(dates)), index=dates)
    ret_arr = ret_d.to_numpy(dtype=float)
    position_sum_arr = np.zeros_like(ret_arr, dtype=float)
    cohort_ret_sum_arr = np.zeros(len(dates), dtype=float)
    active_cohorts_arr = np.zeros(len(dates), dtype=int)
    missing_return_cohorts_arr = np.zeros(len(dates), dtype=int)
    cohort_rows = []

    formation_months = list(opened_weights.index)
    for i, formation_month in enumerate(formation_months):
        entry_date = entry_dates.loc[formation_month]
        w0 = opened_weights.loc[formation_month]
        if pd.isna(entry_date) or np.isclose(w0.abs().sum(), 0.0):
            continue

        exit_date = entry_dates.iloc[i + K] if i + K < len(formation_months) else pd.NaT
        entry_trade_i = int(date_pos.loc[entry_date])
        first_return_i = entry_trade_i + 1
        exit_i = len(dates) if pd.isna(exit_date) else int(date_pos.loc[exit_date]) + 1
        if exit_i <= first_return_i:
            continue

        cohort_rets = ret_arr[first_return_i:exit_i]
        cohort_rets_for_drift = np.nan_to_num(cohort_rets, nan=0.0)
        w0_arr = w0.to_numpy(dtype=float)
        drift = np.cumprod(1.0 + cohort_rets_for_drift, axis=0)
        drift = np.vstack([np.ones((1, drift.shape[1])), drift[:-1]])
        cohort_positions = drift * w0_arr
        missing_held = np.isnan(cohort_rets) & (np.abs(cohort_positions) > 1e-12)
        cohort_daily_ret = np.sum(cohort_positions * cohort_rets_for_drift, axis=1)
        if not fill_missing_prices_while_held:
            cohort_daily_ret[missing_held.any(axis=1)] = np.nan

        active_slice = slice(first_return_i, exit_i)
        position_sum_arr[active_slice] += cohort_positions
        cohort_ret_sum_arr[active_slice] += np.nan_to_num(cohort_daily_ret, nan=0.0)
        active_cohorts_arr[active_slice] += 1
        missing_return_cohorts_arr[active_slice] += np.isnan(cohort_daily_ret).astype(int)

        cohort_rows.append({
            "formation_date": formation_month,
            "entry_date": entry_date,
            "first_return_date": dates[first_return_i],
            "exit_date": exit_date,
            "holding_days": exit_i - first_return_i,
            "winners": int((w0 > 0).sum()),
            "initial_gross": float(w0.abs().sum()),
        })

    active_cohorts = pd.Series(active_cohorts_arr, index=dates, dtype=int)
    position_denominator = np.where(active_cohorts_arr[:, None] > 0, active_cohorts_arr[:, None], np.nan)
    positions_d = pd.DataFrame(position_sum_arr / position_denominator, index=dates, columns=columns).fillna(0.0)
    daily_ret_arr = np.divide(
        cohort_ret_sum_arr,
        active_cohorts_arr,
        out=np.zeros_like(cohort_ret_sum_arr),
        where=active_cohorts_arr > 0,
    )
    if not fill_missing_prices_while_held:
        daily_ret_arr[missing_return_cohorts_arr > 0] = np.nan
    daily_ret = pd.Series(daily_ret_arr, index=dates)

    active = active_cohorts.gt(0)
    daily_ret = daily_ret.loc[active.idxmax():] if active.any() else daily_ret.iloc[0:0]
    monthly_ret = (1 + daily_ret).resample("ME").prod() - 1
    if not fill_missing_prices_while_held:
        monthly_ret = monthly_ret.mask(daily_ret.isna().resample("ME").any())

    diagnostics = pd.DataFrame({
        "eligible": eligible.sum(axis=1),
        "winners": winners.sum(axis=1),
        "members": member_m.sum(axis=1),
        "gross_formed": formed_weights.abs().sum(axis=1),
        "gross_opened": opened_weights.abs().sum(axis=1),
        "entry_date": entry_dates,
        "entry_rejected_no_next_day": formed_weights.where(~entry_possible, 0.0).abs().sum(axis=1),
        "entry_rejected_not_member": formed_weights.where(entry_possible, 0.0).where(~entry_member, 0.0).abs().sum(axis=1),
        "entry_rejected_no_price": formed_weights.where(entry_member & ~entry_price_ok, 0.0).abs().sum(axis=1),
    })

    return {
        "daily_ret": daily_ret,
        "monthly_ret": monthly_ret,
        "daily_equity": (1 + daily_ret).cumprod(),
        "monthly_equity": (1 + monthly_ret).cumprod(),
        "formed_weights": formed_weights,
        "opened_weights": opened_weights,
        "daily_positions": positions_d,
        "daily_stock_returns": ret_d,
        "signal": signal,
        "cohorts": pd.DataFrame(cohort_rows),
        "active_cohorts": active_cohorts,
        "membership_daily": membership_d,
        "diagnostics": diagnostics,
    }
```

```python
def market_returns_from_weights(tr_daily: pd.DataFrame, weights_wide: pd.DataFrame, on_index_m: pd.DataFrame) -> tuple[pd.Series, pd.Series]:
    stock_ret = tr_daily.ffill().pct_change(fill_method=None)
    weights = weights_wide.reindex(index=tr_daily.index, columns=tr_daily.columns).fillna(0.0)
    on_index_d = expand_membership_to_daily(on_index_m, tr_daily.index).reindex(columns=tr_daily.columns).fillna(0).astype(bool)
    weights = weights.where(on_index_d, 0.0)
    weights = weights.div(weights.sum(axis=1).replace(0, np.nan), axis=0).fillna(0.0)
    daily_market_ret = (weights.shift(1) * stock_ret).sum(axis=1, min_count=1).fillna(0.0)
    monthly_market_ret = (1 + daily_market_ret).resample("ME").prod() - 1
    return daily_market_ret, monthly_market_ret

market_daily_ret, market_monthly_ret = market_returns_from_weights(tr_daily, weights_wide, on_index_m)
market_equity = (1 + market_daily_ret).cumprod()
```

```python
bt = winners_only_backtest(
    tr_daily=tr_daily,
    membership_m=membership_m,
    J=6,
    K=6,
    q=0.10,
    skip=1,
    min_names=10,
)

print("Winners-only daily performance")
display(summarize_returns(bt["daily_ret"], periods_per_year=252).to_frame("J6_K6_q10"))

print("Winners-only monthly performance")
display(summarize_returns(bt["monthly_ret"], periods_per_year=12).to_frame("J6_K6_q10"))

display(bt["diagnostics"].tail())
```

```python
fig, ax = plt.subplots(figsize=(11, 5))
bt["daily_equity"].plot(ax=ax, lw=1.8, label="Winners-only")
market_equity.reindex(bt["daily_equity"].index).plot(ax=ax, lw=1.4, label="OSEBX proxy")
ax.set_title("Winners-Only Momentum vs OSEBX Proxy")
ax.set_ylabel("Equity, start = 1")
ax.grid(True, alpha=0.25)
ax.legend()
plt.show()
```

```python
def normal_cdf(x: float) -> float:
    return 0.5 * (1.0 + math.erf(x / np.sqrt(2.0)))


def ols_hac_capm_alpha(
    strategy_monthly_ret: pd.Series,
    market_monthly_ret: pd.Series,
    rf_monthly: float | pd.Series = 0.0,
    lags: int | None = None,
) -> pd.Series:
    """CAPM regression with HAC/Newey-West SEs. Tests H1: alpha > 0."""
    data = pd.concat(
        {
            "strategy": strategy_monthly_ret,
            "market": market_monthly_ret,
        },
        axis=1,
    ).dropna()

    if isinstance(rf_monthly, pd.Series):
        rf = rf_monthly.reindex(data.index).fillna(0.0)
    else:
        rf = pd.Series(float(rf_monthly), index=data.index)

    y = (data["strategy"] - rf).to_numpy(dtype=float)
    x_mkt = (data["market"] - rf).to_numpy(dtype=float)
    X = np.column_stack([np.ones(len(data)), x_mkt])
    n, k = X.shape

    beta = np.linalg.lstsq(X, y, rcond=None)[0]
    resid = y - X @ beta

    if lags is None:
        lags = int(np.floor(4 * (n / 100) ** (2 / 9)))
    lags = int(max(0, min(lags, n - 1)))

    s = np.zeros((k, k))
    for t in range(n):
        xt = X[t:t+1].T
        s += resid[t] ** 2 * (xt @ xt.T)
    for ell in range(1, lags + 1):
        weight = 1.0 - ell / (lags + 1)
        gamma = np.zeros((k, k))
        for t in range(ell, n):
            xt = X[t:t+1].T
            xl = X[t-ell:t-ell+1].T
            gamma += resid[t] * resid[t - ell] * (xt @ xl.T)
        s += weight * (gamma + gamma.T)

    xx_inv = np.linalg.inv(X.T @ X)
    cov = xx_inv @ s @ xx_inv
    se = np.sqrt(np.diag(cov))

    alpha, beta_mkt = beta
    alpha_se = se[0]
    alpha_t = alpha / alpha_se if alpha_se > 0 else np.nan
    p_alpha_gt_0 = 1.0 - normal_cdf(alpha_t) if np.isfinite(alpha_t) else np.nan
    p_alpha_two_sided = 2.0 * (1.0 - normal_cdf(abs(alpha_t))) if np.isfinite(alpha_t) else np.nan
    r2 = 1.0 - np.sum(resid ** 2) / np.sum((y - y.mean()) ** 2)

    return pd.Series({
        "N months": n,
        "NW lags": lags,
        "Alpha monthly": alpha,
        "Alpha annualized": alpha * 12,
        "Alpha HAC SE": alpha_se,
        "Alpha t-stat": alpha_t,
        "p-value alpha > 0": p_alpha_gt_0,
        "p-value alpha != 0": p_alpha_two_sided,
        "Market beta": beta_mkt,
        "R2": r2,
    })

RISK_FREE_ANNUAL = 0.03
RF_MONTHLY = (1.0 + RISK_FREE_ANNUAL) ** (1.0 / 12.0) - 1.0

capm_baseline = ols_hac_capm_alpha(
    strategy_monthly_ret=bt["monthly_ret"],
    market_monthly_ret=market_monthly_ret,
    rf_monthly=RF_MONTHLY,
    lags=max(0, 6 - 1),
)

print("CAPM alpha test: H0 alpha = 0, H1 alpha > 0")
display(capm_baseline.to_frame("J6_K6_q10"))
```

```python
def run_winners_grid(
    tr_daily: pd.DataFrame,
    membership_m: pd.DataFrame,
    market_monthly_ret: pd.Series,
    Js=(3, 6, 9, 12),
    Ks=(1, 3, 6, 9, 12),
    qs=(0.05, 0.10, 0.20),
    skip: int = 1,
    min_names: int = 10,
) -> pd.DataFrame:
    rows = []
    for J in Js:
        for K in Ks:
            for q in qs:
                bt_i = winners_only_backtest(
                    tr_daily=tr_daily,
                    membership_m=membership_m,
                    J=J,
                    K=K,
                    q=q,
                    skip=skip,
                    min_names=min_names,
                )
                perf = summarize_returns(bt_i["monthly_ret"], periods_per_year=12)
                capm = ols_hac_capm_alpha(
                    strategy_monthly_ret=bt_i["monthly_ret"],
                    market_monthly_ret=market_monthly_ret,
                    rf_monthly=RF_MONTHLY,
                    lags=max(0, K - 1),
                )
                rows.append({"J": J, "K": K, "q": q, "skip": skip, **perf.to_dict(), **capm.to_dict()})
    return pd.DataFrame(rows).sort_values("p-value alpha > 0").reset_index(drop=True)

winners_grid = run_winners_grid(tr_daily, membership_m, market_monthly_ret)
display(winners_grid)
```

```python
best = winners_grid.iloc[0]
best_J = int(best["J"])
best_K = int(best["K"])
best_q = float(best["q"])

bt_best = winners_only_backtest(
    tr_daily=tr_daily,
    membership_m=membership_m,
    J=best_J,
    K=best_K,
    q=best_q,
    skip=1,
    min_names=10,
)

print("In-sample best by CAPM alpha p-value. Treat as exploratory, not out-of-sample evidence.")
display(best.to_frame("best_in_sample"))

fig, ax = plt.subplots(figsize=(11, 5))
bt_best["daily_equity"].plot(ax=ax, lw=1.8, label="Best winners-only")
market_equity.reindex(bt_best["daily_equity"].index).plot(ax=ax, lw=1.4, label="OSEBX proxy")
ax.set_title(f"In-Sample Best Winners-Only: J={best_J}, K={best_K}, q={best_q:.0%}")
ax.set_ylabel("Equity, start = 1")
ax.grid(True, alpha=0.25)
ax.legend()
plt.show()

latest_weights = bt_best["opened_weights"].loc[bt_best["opened_weights"].abs().sum(axis=1).gt(0)]
latest_date = latest_weights.index.max()
latest_winners = (
    latest_weights.loc[latest_date]
    .loc[lambda s: s > 0]
    .sort_values(ascending=False)
    .rename("weight")
    .reset_index()
    .rename(columns={"ticker": "company"})
)
latest_winners.insert(0, "formation_date", latest_date)
print("Latest opened winners portfolio")
display(latest_winners)
```

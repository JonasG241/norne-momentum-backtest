Bruk CSV fil for total return, ikke PX. Total return er medregnet reinvestert dividende

Dette gir et features df som inneholder tr (pris + reinvestiret dividende), mva50, mva100, mva200, i et wide format, som betyr dobbel header.
Weights er et df med daglig forhold mellom market cap av hvert firma.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
path_total_return = './data/raw/TOTRET_DAILY.csv' # TOTAL RETURN DAILY CSV. endre path slik at det blir riktig for deg. 
path_market_cap = './data/raw/MCAP_DAILY.csv' # MARKET CAP DAILY CSV. endre path slik at det blir riktig for deg. 
```

```python


df_totret = pd.read_csv(path_total_return, sep=';', decimal=',', parse_dates=['Date'], dayfirst=True, na_values=['#N/A'])
df_totret.set_index('Date', inplace=True)
```

```python
# Read market cap and compute per-date portfolio weights

df_mcap = pd.read_csv(path_market_cap, sep=';', decimal=',', parse_dates=['Date'], dayfirst=True, na_values=['#N/A'])
# Use Date as index
df_mcap.set_index('Date', inplace=True)
# Ensure numeric values (coerce malformed entries to NaN)
df_mcap = df_mcap.apply(pd.to_numeric, errors='coerce')
# Total market cap per date
total_mcap = df_mcap.sum(axis=1)
# Weights: each asset mcap divided by total mcap on the same date
weights = df_mcap.div(total_mcap, axis=0)
# Inspect the first rows of weights
weights.head()
```

```python
mva50  = df_totret.rolling(50,  min_periods=1).mean()
mva100 = df_totret.rolling(100, min_periods=1).mean()
mva200 = df_totret.rolling(200, min_periods=1).mean()
```

```python
features = pd.concat(
    {"tr": df_totret, "mva50": mva50, "mva100": mva100, "mva200": mva200},
    axis=1
)
# access like:
# features["mva200"]["EQNR.OL"]
```
